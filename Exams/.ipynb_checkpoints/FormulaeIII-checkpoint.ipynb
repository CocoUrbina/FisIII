# RLC

## RL

$U_T = U_L + U_C$

$\displaystyle U_L = \frac{1}{2} LI^2$

$\displaystyle U_C = \frac{1}{2}CV^2 = \frac{1}{2} \frac{q^2}{C}$

$\displaystyle \frac{d U_T}{dt} = LI\frac{dI}{dt} + \frac{q}{C} \frac{dq}{dt} = 0$

$I = qt$ $I = \frac{dq}{dt}$ ??

$\boxed{q = q_m \cos(\omega t + \phi)}$,  $\omega = \sqrt{1/LC}$

$U_T = \frac{1}{2C} q_m^2 = cte$

## RLC

$\displaystyle \frac{dU}{dt} = -RI^2$, por la resistencia.

$\displaystyle \frac{d U_T}{dt} = LI\frac{dI}{dt} + \frac{q}{C} \frac{dq}{dt} = -RI^2$

- $R/2L > \omega$: $q = A_1 \exp(s_1t) + A_2 \exp(s_2t)$, sobreamortiguado
- $R/2L = \omega$: $q = \exp(-Rt/2L) (q_m + At)$, amortiguado critico
- $R/2L < \omega$: $q = \exp(-Rt/2L) \cos(w^\prime t + \phi)$, subamortiguado

**Oscilaciones forzadas**

$F = F_m \cos(\omega^{\prime \prime} t)$

$I = I_m \sin(\omega^{\prime \prime} t - \phi)$
