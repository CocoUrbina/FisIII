### Ejercicio #1
En el circuito mostrado, las bobinas con inductancia $L_2$ y $L_3$ están conectadas en serie y acopladas magnéticamente con una inductancia mutua $M_{23}$. Las bobinas con inductancias $L_1$ y $L_4$ estan acopladas en paralelo y acopladas entre sí con una inductancia mutua $M_{14}$. No existen otros acoplamientos magnéticos entre las bobinas.

Determine la inductancia equivalente total. Considere el resultado del signo para los dos tipos de acoplamiento posibles: positivos (los flujos magnéticos se refuerzan) y negativo (los flujos magnéticos se oponen).

<center>
  <img src="Ej_S14_01.jpg" style="width:40%" />
</center>


### Solución

$L_{23} = L_2 + L_3 + 2\sigma_{23}M_{23}$

con $\sigma_{23} = +1$ si se refuerzas o $\sigma_{23} = -1$ si no se refuerzan.


$\displaystyle L_{14} = \frac{L_1 L_4 - M_{14}^2}{L_1 + L_4 - 2\sigma_{14} M_{14}}$

Ahora toca hacer $L_{23} \parallel L_{14}$

$\displaystyle \frac{1}{L_{eq}} = \frac{1}{L_{23}} + \frac{1}{L_{14}} $

$\displaystyle L_{eq} = \frac{L_{23}\times L_{14}}{L_{23} + L_{14}}$

**Nota**

- En serie:

Va ---- L2 ---- Vb ---- L3 ---- Vc

$V_a - V_b = L_2 \frac{dI}{dt} + \sigma_{23}M_{23} \frac{dI}{dt}$

$V_b - V_c = L_3 \frac{dI}{dt} + \sigma_{23}M_{23} \frac{dI}{dt}$

sumando ambas:

$V_a - V_c = (L_2 + L_3) \frac{dI}{dt} + 2\sigma_{23}M_{23} \frac{dI}{dt}$

Entonces $L_{eq} = (L_2 + L_3)$


- En paralelo:

$V = L_1 \frac{dI_1}{dt} + \sigma_{14}M_{14} \frac{dI_4}{dt}$

$V = L_4 \frac{dI_4}{dt} + \sigma_{14}M_{14} \frac{dI_1}{dt}$

$\displaystyle \mathbb{V} = \mathbb{L_{14}} \frac{d \mathbb{I}}{dt}$, con $\mathbb{I} = \begin{pmatrix}
I_1 \\
I_4
\end{pmatrix} $

$
\begin{pmatrix}
V \\
V
\end{pmatrix}
= \begin{pmatrix}
L_1 & \sigma_{14}M_{14} \\
\sigma_{14}M_{14} & L_4
\end{pmatrix}
\begin{pmatrix}
\frac{dI_1}{dt} \\
\frac{dI_4}{dt}
\end{pmatrix}
$

$\text{det}(\mathbb{L}) = L_1 L_4 - M_{14}^2$

$V = L_{eq} \frac{d I_T}{dt}$, con $I_T = I_1 + I_4$

$\mathbb{1}
=\begin{pmatrix}
1 \\
1
\end{pmatrix}$

$\displaystyle V = \frac{1}{\mathbb{1}^{T}  \mathbb{L}^{-1} \mathbb{1}} \frac{d I_T}{dt}$

**Prueba de** $\displaystyle V = \frac{1}{\mathbb{1}^{T}  \mathbb{L}^{-1} \mathbb{1}} \frac{d I_T}{dt}$

Considere que $I_T = \mathbb{1}^T \mathbb{I}$, $V\mathbb{1} = \mathbb{V}$ 

Vea que

$\displaystyle \frac{d \mathbb{I} }{dt} = \mathbb{L}^{-1} V \mathbb{1} $

$\displaystyle \frac{d \mathbb{I} }{dt} = V \mathbb{1}^T \mathbb{L}^{-1} \mathbb{1} $

$\displaystyle V = \frac{1}{\mathbb{1}^{T}  \mathbb{L}^{-1} \mathbb{1}} \frac{d I_T}{dt}$

### Ejercicio #2
En un circuito LC amortiguado, ¿cuál es el tiempo requerido tal que la energía máxima en el capacitor durante una oscilación disminuya a la mitad de la energía máxima inicial? Suponga $q=q_{max}$ en $t=0$.

### Solución

$U_{max}$

$Q \propto \exp \left( \frac{-R}{2L} t \right) $

$U \propto \exp \left( \frac{-R}{L} t \right) $

$Q = A \exp \left( \frac{-R}{L} t \right) \cos(\omega t + \phi)$, $\phi$ y $A$ condiciones de frontera.

$Q = A \cos(\phi) = Q_m$

$A = Q_m$, $\phi = 0$


$\frac{1}{2} U(t_1) = U(t_1+t_2)$

$\exp \left( \frac{-R}{L} t_2 \right) = \frac{1}{2}  \rightarrow t_2 = \frac{L}{R} \ln(2)$.

### Ejercicio #3
Determine las corrientes en cada una de las ramas.

<center>
  <img src="Ej_S14_03.jpg" style="width:50%" />
</center>


V = 1 A

$\bar{X}_c = 12$

$\bar{X}_L = 8$

$R = 10$

### Solución

$\displaystyle I_R = \frac{\mathcal{E}}{R}$

$\displaystyle I_L = \frac{\mathcal{E}}{\bar{X}_L}$

$\displaystyle I_C = \frac{\mathcal{E}}{\bar{X}_C}$

Necesitamos $\mathcal{E}$

$\mathcal{E} = I* Z$

Necesitamos $Z$


$\displaystyle \frac{1}{Z} = \frac{1}{-j X_c} + \frac{1}{j X_L} + \frac{1}{R}$

$\displaystyle \frac{1}{Z} = j \left(\frac{1}{X_c} - \frac{1}{X_L} \right) + \frac{1}{R}$

$\displaystyle \frac{1}{Z} = - \frac{1}{24}j  + \frac{1}{10}$

$\displaystyle Z = \frac{240}{-10j + 24} * \frac{10j + 24}{10j + 24} = 3.55j + 8.52$

$Z = 9.23 \Omega < 22.6 \degree$

$\mathcal{E} = I* Z = (-1,0 \degree)(9.23,22.6 \degree) = (-9.23, 22.6 \degree)$

$\displaystyle I_R = \frac{\mathcal{E}}{R} = \frac{(-9.23, 22.6 \degree)}{(10, 0 \degree)} = (0.92, 202.6 \degree) $

$\displaystyle I_L = \frac{\mathcal{E}}{\bar{X}_L} = (1.15, 112.3 \degree)$

$\displaystyle I_C = \frac{\mathcal{E}}{\bar{X}_C} = (0.77, 292.3 \degree)$


### Ejercicio #4
Determine la corriente a través del capacitor.

$I_1 = 10\times 10^{-2}A < 0\degree$

$I_2 = 3\times 10^{-3}A < 90\degree$

<center>
  <img src="Ej_S14_04.jpg" style="width:50%" />
</center>


### Solución

$\bar{I}_1 = 10^{-2}A < 0\degree$

$\bar{I}_2 = 3\times 10^{-3}A < 90\degree$

$\bar{X}_C = -j2k$

$\bar{X}_{L1} = j1k$

$\bar{X}_{L2} = j2.5k$

$R=600 \Omega$

Luego,

$I_1  = I_1^\prime + I_C$ 

$I_1^\prime$ es la que sigue directo por arrriba

$I_C$ es la que baja hacia el capacitor

Luego, 

$I_3  = I_1^\prime + I_2$ 

$I_3$ es la que sigue por arriba

$I_1^\prime$ es la que viene por arriba

Combinando ambas,

$I_C = I_1 + I_2 - I_3$

$\mathcal{E} = I_C X_C = I_3 Z^\prime$

$I_C X_C = (I_1 + I_2 - I_C) Z^\prime$

$\displaystyle I_C = \frac{(I_1 + I_2) Z^\prime}{X_C + Z^\prime}$

$\displaystyle Z^\prime = \frac{RjX_{L2}}{R+jX_{L2}} + jX_{L1} = \left( \frac{R^2 X_{L2}}{R^2 + X_{L2}^2} + X_{L1} \right)j + \frac{RX_{L2}^2}{R^2 + X_{L2}^2}$

$Z^\prime = 1136.2j + 567.3 = (1269.9\Omega, 63.5\degree)$

Ahora,

$I_1 + I_2 = 10^{-2} + 3\times10^{-3}j = (10.4\times10^{-3}A, 16.7\degree)$

$X_C + Z^\prime = 567.3 -j865.3 = (1034 \Omega, -56.8\degree)$

$\displaystyle I_C = \frac{(10.4\times10^{-3}A, 16.7\degree) (1269.9\Omega, 63.5\degree) }{(1034 \Omega, -56.8\degree)} = (12.8 mA, 137\degree)$
