### Ejercicio #3: HW04, P04 Ampere

Considere un cable coaxial compuesto por un centro aislante, un conductor cilı́ndrico de radio interno $R_1$ y radio externo $R_2$ . Otro aislante, de radio externo $R_3$ , evita el cortocircuito con un último conductor de radio externo $R_4$ y todo cubierto por un aislante final. En determinado momento, el conductor interno lleva una densidad de corriente variable de la forma $\displaystyle \vec{j} = \frac{\alpha I}{r}$ y el conductor externo lleva una densidad de corriente en sentido contrario de la forma $\displaystyle \vec{j} = -\left( \frac{\alpha_1 I}{r^2} + \frac{\alpha_2 I}{r^3} \right)$. Determine el campo magnético utilizando la ley de Ampere en cada una de las zonas: 

- Zona 1: $r < R_1$
- Zona 2: $R_1 < r < R_2$
- Zona 3: $R_2 < r < R_3$
- Zona 4: $R_3 < r < R_4$
- Zona 5: $R_4 < r$

### Ejercicio #4: Exa02 P02 Ampere

Considere un cable coaxial compuesto por un centro conductor cilı́ndrico de radio interno $R_1$. Un aislante, de radio externo $R_2$ , evita el cortocircuito con un último conductor de radio externo $R_3$ y todo cubierto por un aislante final. En determinado momento, el conductor interno conduce de manera uniforme una corriente $I_1$ en una dirección dada. Si el conductor externo lleva una densidad de corriente en sentido contrario a $I_1$, y descrito por $\displaystyle \vec{j_2} = - \left( \frac{\alpha_1 I}{r^2} \right)$:

1. Determine el campo magnético utilizando la ley de Ampere en cada una de las zonas: 

- Zona 1: $r < R_1$
- Zona 2: $R_1 < r < R_2$
- Zona 3: $R_2 < r < R_3$
- Zona 4: $R_3 < r$

2. Determine el radio en el que el campo magnético debe anularse.
