# Movimiento de partículas cargadas en campos eléctricos no uniformes

## Motivación

Nos interesa el estudio de la dinámica de una partícula cargada en un campo eléctrico no uniforme. En particular, estudiamos la geometria de un anillo con carga.

Nuestra presentación es demostrativa mediante herramientas computacionales.

## Problema

El Resnick (pp.599-600) [3] plantea el  estudio del movimiento de una partícula cargada, lanzada a lo largo del eje $z$ hacia el anillo con una velocidad inicial $v_0$. Se desprecian los efectos de la gravedad y se considera únicamente la fuerza eléctrica.

En particular, el Resnick plantea la discusión de un protón con velocidad inicial $v_0 = -7 \times 10^{5} ~\mathrm{m/s}$ y un anillo cuya densidad de carga lineal es $\lambda = 2 \times 10^{-7} ~\mathrm{C/m}$ y cuyo radio es $R = 3~\mathrm{cm}$.

<img src="Fig1.png" style="width:100%; max-width:900px;"/>

Para un anillo con carga linea uniforme, la magnitud del campo eléctrico a lo largo de su eje axial

$$E_z = \frac{\lambda}{2\epsilon_0} \frac{Rz}{(z^2 + R^2)^{3/2}} \tag{Resnick ec. 26-18}$$

Con $\displaystyle \lambda = \frac{Q}{s} = \frac{Q}{2 \pi R}$ y $\displaystyle k = \frac{1}{4 \pi \epsilon_0}$, 

$$\boxed{ E_z = \frac{kQz}{(z^2 + R^2)^{3/2}}} \tag{1}$$

### Ecuación de movimiento
De la ecuación $F_e = qE$, tenemos que

$$ \boxed{m\ddot{z} = qkQ  \frac{z}{(z^2 + R^2)^{3/2}}} \tag{2}$$

La ecuación diferencial anterior **no es lineal** debido a la potencia $3/2$, por lo que no tiene solución analítica.

## Metodología

Se emplean métodos numéricos, dividiendo el movimiento en intervalos infinitesimalmente pequeños, suponiendo que la aceleración es casi constante, obteniendo una solución aproximada.

Se implementan el método de Euler y/o Runge-Kutta 4 para actualizar los valores de la posición $z(t)$ y la velocidad $v(t)$ en función de sus valores anteriores.

### Método de Euler

De [1]: 


>El método de Euler es muy sencillo, se basa en la expansión de Taylor de la función $z(t)$. Tenemos
$$
\text{Expansión Taylor} \Rightarrow z(t+h) = z(t) + h\frac{dz}{dt} + \overbrace{ \frac{h^2}{2} \frac{d^2z}{dt^2} } ^{\epsilon} + O(h^3).
$$
>Esto implica que para avanzar en el tiempo la función por un paso $h$, el cual suponemos que es lo suficientemente pequeño, basta con utilizar la ecuación
$$
\boxed{z(t + h) = z(t) + hf(z,t).} \tag{3}
$$
>[...]
>* El algoritmo toma la siguiente forma:
>  - Empezar con $t = t_0$, $z = z_0$
>  - Discretizar el tiempo en pasos temporales de forma equidistante con espaciamiento $h$, donde cada punto en el tiempo está denotado con $t_i$
>  - Para cada punto en el tiempo encontrar $z$ utilizando el resultado de la iteración previa: $z_i = z_{i-1} + hf(z_{i-1})$

### Método de Runge-Kutta 4

La actualizacion es

$$
\boxed{z(t+h) = z(t) + \frac{1}{6}(k_1 + 2 k_2 + 2k_3 + k_4)} \tag{4}
$$ 

Donde,

* $k_1 = hf(z, t)$,
* $k_2 = hf\left(z + \frac{k_1}{2}, t+\frac{h}2\right)$,
* $k_3 = hf\left(z + \frac{k_2}{2}, t+\frac{h}2\right)$,
* $k_4 = hf\left(z + k_3, t + h \right)$.


## Implementación

### Librerias

```python
# Libraries
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from ipywidgets import interact, FloatSlider, VBox
from mpl_toolkits.mplot3d import Axes3D
import warnings
warnings.filterwarnings("ignore")
```

### Constantes físicas

```python
# Physics constants  
eps0 = 8.8541878128e-12    # vacuum permittivity  [C^2 / N·m^2] 
k = 1 / (4 * np.pi * eps0) # [N·m^2 / C^2]
```

### Parámetros del problema

```python
# Particle (proton)
q =  1.602176634e-19        # proton charge [C]
m = 1.67262192369e-27       # proton mass [kg]

# Particle (electron)
#q =  -1.602176634e-19      # electron charge [C]
#m = 9.1093837015e-31       # electron mass [kg] 

# Ring
R = 0.03                      # ring's radius [m]
lamb = 2e-7                   # linear density charge [C/m]
Q =  lamb * (2 * np.pi * R)   # total ring's charge [C]
```

### Campo Eléctrico

```python
def Ez(z):
    return k * Q * z / (R**2 + z**2)**(3/2)  # Electric field along the z-exe due to the ring [N/C]
```

### Método de Euler

```python
def euler(z0, v0, t_total, h=5e-10):
    pasos = int(t_total / h)
    z = np.zeros(pasos) # lista de estados z(t_i)
    v = np.zeros(pasos) # lista de estados v(t_i)
    t = np.linspace(0, t_total, pasos) # t_i

    z[0] = z0 # valor inicial (se pasan desde el slider)
    v[0] = v0 # valor inicial (se pasan desde el slider)

    for i in range(1, pasos): #Euler
        a = q * Ez(z[i-1]) / m # aceleración : a = F/m = qE/m
        v[i] = v[i-1] + a * h # actualización de v en función del v anterior (Euler explícito)
        z[i] = z[i-1] + v[i] * h # actualización de z en función del z anterior (Euler explícito)

        if abs(z[i]) > 0.5: #rango de estudio para el desplazamiento [-0.5, 0.5] (condición de parada)
            return t[:i], z[:i], v[:i]

    return t, z, v
```

### Método de Runge-Kutta 4

```python
def rk4(z0, v0, t_total, h=5e-9): # paso más grande con respecto a Euler
    pasos = int(t_total / h)
    z = np.zeros(pasos)
    v = np.zeros(pasos)
    t = np.linspace(0, t_total, pasos)

    z[0] = z0 
    v[0] = v0

    # ¡EDOs de orden 1 acopladas!
    # dz/dt = v
    # dv/dt = a(z) = q*Ez(z) / m
    def f_z(z, v):
        return v

    def f_v(z):
        return q * Ez(z) / m

    for i in range(1, pasos): #RK4
        # k1
        k1_z = f_z(z[i-1], v[i-1])
        k1_v = f_v(z[i-1])

        # k2
        k2_z = f_z(z[i-1] + 0.5*h*k1_z, v[i-1] + 0.5*h*k1_v)
        k2_v = f_v(z[i-1] + 0.5*h*k1_z)

        # k3
        k3_z = f_z(z[i-1] + 0.5*h*k2_z, v[i-1] + 0.5*h*k2_v)
        k3_v = f_v(z[i-1] + 0.5*h*k2_z)

        # k4
        k4_z = f_z(z[i-1] + h*k3_z, v[i-1] + h*k3_v)
        k4_v = f_v(z[i-1] + h*k3_z)

        # Actualización RK4
        z[i] = z[i-1] + (h/6)*(k1_z + 2*k2_z + 2*k3_z + k4_z)
        v[i] = v[i-1] + (h/6)*(k1_v + 2*k2_v + 2*k3_v + k4_v)

        # Condición de parada
        if abs(z[i]) > 0.5:
            return t[:i], z[:i], v[:i]

    return t, z, v
```

### Simulación y gráficos

```python
from IPython.display import HTML

def modelar(z0, v0, t_total):

    # Simulación
    t, z, v = rk4(z0, v0, t_total) # Invoquemos mejor RK4. Euler toma más tiempo.
    n = len(z)

    # Figura completa 
    fig = plt.figure(figsize=(10, 10))

    # Grafico 1:  z(t) vs t
    ax1 = fig.add_subplot(2, 2, 1) # 2x2, 1 = arriba izq.
    ax1.plot(t*1e7, z, lw=2) 
    ax1.set_xlabel("t ($×10^{-7}$ s)")
    ax1.set_ylabel("z (m)")
    ax1.grid(True)
    punto_z, = ax1.plot([t[0]*1e7], [z[0]], 'ro', ms=8) # Esto extrae la línea del ax devuelto por plot, no la lista.

    # 2) Gráfico 2: v(t) vs t
    ax2 = fig.add_subplot(2, 2, 2) # 2x2, 2= arriba derecha
    ax2.plot(t*1e7, v/1e6, lw=2)
    ax2.set_xlabel("t ($×10^{-7}$ s)")
    ax2.set_ylabel("$v_z$ (×$10^6$ m/s)")
    ax2.grid(True)
    punto_v, = ax2.plot([t[0]*1e7], [v[0]/1e6], 'ro', ms=8)

    # 3) Gráfico 3: 3D Animation
    ax3 = fig.add_subplot(2, 2, (3,4), projection='3d') # 2x2, (3,4)= abajo izq. y der.
    ax3.set_xlim([-0.35, 0.35])
    ax3.set_ylim([-0.35, 0.35])
    ax3.set_zlim([-0.55, 0.55])
    
    # Ring
    theta = np.linspace(0, 2*np.pi, 200) #200 ángulos uniformes de 0 a pi
    ax3.plot(R*np.cos(theta), R*np.sin(theta), 0, color='orange', linewidth=3)
    
    # Particle
    punto3d, = ax3.plot([0], [0], [z[0]], 'ro', markersize=8)
    ax3.set_title("Movimiento 3D de la particula")

    # Función de actualización
    def update(i): # los i son los del arreglo frames
        punto_z.set_data([t[i]*1e7], [z[i]]) # actualiza el punto rojo en z vs t
        punto_v.set_data([t[i]*1e7], [v[i]/1e6]) # actualiza el punto rojo en v vs t
        punto3d.set_data([0], [0]) # actualiza el punto rojo en el 3D en x, y (siempre ceros)
        punto3d.set_3d_properties(z[i]) # actualiza la coordena z del punto rojo en el 3D
        return punto_z, punto_v, punto3d # devuelve el punto actualizado

    anim = FuncAnimation(
        fig,
        update, # la función anterior que mueve los puntos
        frames=range(0, n, 10),  # 1 de cada 10 frames. Los que recibe update
        interval=100, # tiempo entre los frames
        blit=False # redibuja todo en cada frame
    )

    plt.close(fig)   # evita doble render

    return HTML(anim.to_jshtml())
```

```python
# Sliders
interact(
    lambda z0, v0, t_total: modelar(z0, v0, t_total),

    z0=FloatSlider(
        value=0.5, min=-0.5, max=0.5, step=0.001,
        description="z₀ (m)",
        layout={'width': '800px'}
    ),

    v0=FloatSlider(
        value=-7e5, min=-2e6, max=2e6, step=1e2,
        description="v₀ (m/s)",
        layout={'width': '800px'}
    ),

    t_total=FloatSlider(
        value=15e-7, min=1e-8, max=3e-3, step=1e-8,
        description="t total (s)",
        layout={'width': '800px'}
    ),
)
```

### Caso #3: Electrón - Movimiento Anharmónico

Consideremos ahora un electrón. En este caso la fuerza eléctrica es de atracción, por lo que ahora la partícula describirá un movimiento anharmónico ($F_e$ no es lineal).

$z_0 = 0.05~\mathrm{m}$

$v_0 = -7.0 \times 10^{5}~\mathrm{m/s}$

$q \rightarrow -q$

### Caso #4: Electrón - Movimiento Armónico Simple


Consideremos un desplazamiento inicial $z_0$ muy pequeño en relación con el radio del anillo, es decir, $z\ll R$. Retomemos la ecuación del movimiento:
$$ m\ddot{z} = qkQ  \frac{z}{\left( z^2 + R^2 \right)^{3/2}} $$

Si  $z\ll R \ \rightarrow\  z^2 + R^2 \approx R^2  \ \rightarrow\ \left(z^2 + R^2\right)^{3/2} \approx \left( R^2 \right)^{3/2} = R^3$.

Entonces la ecuación del movimiento se puede aproximar como

$$ \boxed {m\ddot{z} \approx qkQ  \frac{z}{R^3}} \tag{5}$$


Nótese que ahora el lado derecho de la ec. (5), es decir, la fuerza restauradora, es una fuerza de Hooke. ($F_e(z) = -|C| z$). Esto debe bastarnos para saber que para oscilaciones en las vecindades del centro del anillo la partícula describirá un movimiento oscilatorio casi armónico.

Reescribiendo la ec.(5) en la forma estándar para osciladores:

$$
m \ddot{z} - qkQ \frac{1}{R^{3}}\, z \approx 0 
\;\;\Longleftrightarrow\;\;
\ddot{z} - \left(\,\frac{qkQ}{mR^{3}}\right) z \approx 0.
$$

Si hacemos $\displaystyle \omega^{2} = \,\frac{qkQ}{mR^{3}}$,

$$
\boxed {\ddot{z} - \omega^{2} z \approx 0 } \tag{6}
$$

que es la ecuación del M.A.S.

$z_0 = 0.01~\mathrm{m}$

$R=0.2~\mathrm{m}$

$v_0 = -7.0 \times 10^{4}~\mathrm{m/s}$

$q \rightarrow -q$
