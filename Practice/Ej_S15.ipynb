### Ejercicio #1
Un capacitor se carga por medio de una corriente que circula a través de un alambre recto y delgado que va desde el centro de una de las placas circulares hasta el centro de la otra. A medida que el capacitor se carga, el campo eléctrico entre las placas aumenta y, por lo tanto, también lo hace la densidad de energía electromagnética almacenada en esa región. Esto implica que debe existir un flujo de enegía que ingrese al capacitor desde una fuente (por jemplo, una batería conectada al alambre).

Demuestre, utilizando el vector de Poynting, que el flujo de energía electromagnética que sale radialmente del alambre es igual a la rapidez del cambio de energía almacenada en el campo eléctrico del capacitor.

$\displaystyle \int \vec{S}\cdot \vec{dA} = \frac{\partial U}{\partial t} = \epsilon_0 E \frac{\partial E}{\partial t} V $

$\displaystyle U = \frac{1}{2} \epsilon_0 E^2 V$


### Solución

$\displaystyle \int \vec{S}\cdot \vec{dA} = \frac{1}{\mu_0} EB$


$\displaystyle \int \vec{B}\cdot \vec{ds} = \mu_0 I - \epsilon_0 \mu_0 A \frac{\partial E}{\partial t} $

$\displaystyle E = \frac {\sigma}{\epsilon_0}$

$\displaystyle I = \frac{dQ}{dt} = \frac{d (\sigma A)}{dt} = \frac{d (\sigma A)}{dt} = \frac{d (\epsilon_0 E A)}{dt} = \epsilon_0   \pi b^2 \frac{d( E)}{dt}$


$\displaystyle B(2 \pi r) = \mu_0 \epsilon_0   \pi b^2 \frac{\partial E}{\partial t} - \epsilon_0 \mu_0 \pi r^2 \frac{\partial E}{\partial t} $

$\displaystyle B = \frac{\mu_0 \epsilon_0}{2} \frac{\partial E}{\partial t} \left(  \frac{b^2 - r^2}{r}  \right)$

$\displaystyle \int \vec{S}\cdot \vec{dA} =\left( \frac{1}{\mu_0} EB \right) (2\pi r l) = 
\epsilon_0 E \frac{\partial E}{\partial t} \left( \pi b^2 - \pi r^2  \right)l
$
