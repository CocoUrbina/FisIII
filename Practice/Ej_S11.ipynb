### Ejercicio #4 - Resnick Ej.34-23, Serway, 31-P62
Una espira rectangular de dimensiones $l$ y $w$ se mueve con una velocidad constante $\vec{v}$ alejándose de un alambre recto e infinito que transporta una corriente $I$ en el plano de la espira, como se muestra en la figura 4. Sea $R$ la resistencia total de la espira. ¿Cuál es la corriente en la espira en el instante en que el lado más cercano se encuentra a una distancia $r$ del alambre?

<center>
  <img src="Ej_S11_04.png" style="width:25%" />
</center>

$\mathcal{E}_{ind} = I_{ind}R$

$ I_{ind} = \mathcal{E}_{ind} / R$ , (horario)

$ \mathcal{E}_{ind} = -d \Phi_B /dt $ , (Faraday)

$ \Phi_B = \int \vec{B}\cdot d\vec{A}$

El campo de la barra infinita: $\displaystyle B = \frac{\mu_0 i}{2 \pi d}$

El campo magnetico entra en la espira perpendicular al plano de la espira. Entonces se simplica a una integral escalar. Dado que la espira es rectangular, $dA = dldr$.

$\displaystyle \Phi_B = \int_r ^{r+w} \int_o ^l \frac{\mu_0 i}{2 \pi r}\ \  dldr$

$\displaystyle \Phi_B = \frac{\mu_0 il}{2 \pi}\ln\left( \frac{r+w}{r} \right)$

$\displaystyle \mathcal{E}_{ind} = - \frac{\mu_0 il}{2 \pi} \frac{d}{dt} \left[ \ln\left( \frac{r+w}{r} \right) \right] $ 

$\displaystyle \mathcal{E}_{ind} = - \frac{\mu_0 il}{2 \pi} \left(  \frac{1}{\frac{r+w}{r}} \right) \left( \frac{-w}{r^2} \right) \dot{r} $ 

$\displaystyle I_{ind} =  \frac{\mu_0 ilv}{2 \pi R} \left( \frac{w}{r(r+w)} \right) $


### Ejercicio #5 - Resnick Pr.34-14
Un campo magnético uniforme $\vec{B}$ llena un volumen cilíndrico de radio $R$. Una varilla metálica de longitud $L$ se coloca como se muestra en la figura 5. Si $B$ varía con una rapidez $dB/dt$, demuestre que la fuerza electromotriz generada por el campo magnético variable que actúa entre los extremos de la varilla está dada por

$$\varepsilon = \frac{dB}{dt}\frac{L}{2} \sqrt{R^2- \left( \frac{L}{2} \right)^2}$$

<center>
  <img src="Ej_S11_05.png" style="width:25%" />
</center>

$ \mathcal{E} = \oint \vec{E}\cdot d\vec{s}$

$ \mathcal{E}_{ind} = -d \Phi_B /dt $ , (Faraday)

$ \oint \vec{E}\cdot d\vec{s} = -d \Phi_B /dt$

$ \vec{E}\cdot d\vec{s} = E (2\pi r)$

$ \Phi_B = \int \vec{B}\cdot d\vec{A} = B(\pi r^2)$

$ E (2\pi r) = (\pi r^2) dB/dt $

$\displaystyle E = \frac{r}{2} dB/dt $, perpendicular a las lineas radiales.

Sea $h$ la distancia del centro del circulo al centro de la barra.

$ \mathcal{E} = \int \vec{E}\cdot d\vec{s}$

$\displaystyle \mathcal{E} = \frac{dB}{dt} \int \frac{r}{2} \frac{h}{r}\ \  dx$ , h/r es seno o coseno

$\displaystyle \mathcal{E} = \frac{dB}{dt} \frac{L}{2} h$

donde $h = R^2 - (L/2)^2$

$\displaystyle \mathcal{E} = \frac{dB}{dt}\frac{L}{2} \sqrt{R^2- \left( \frac{L}{2} \right)^2}$
